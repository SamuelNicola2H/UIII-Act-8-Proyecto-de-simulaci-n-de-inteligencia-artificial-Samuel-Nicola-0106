```python
# ============================================================
# IA PARA MINERO Y PICO
# Uso de if, elif, else y while True
# Nivel: Principiante
# CBTIS 128
# Alumno: Samuel Nicola
# Inteligencia Artificial
# ============================================================

print("================================================")
print("⛏️ IA PARA MINERO Y PICO ⛏️")
print("================================================")
print("Bienvenido al simulador de inteligencia artificial")
print("================================================")

while True:

    print("\n----------- MENÚ PRINCIPAL -----------")
    print("1. Revisar estado del pico")
    print("2. Buscar minerales")
    print("3. Revisar energía del minero")
    print("4. Salir")

    opcion = input("Selecciona una opción: ")

    # OPCIÓN 1
    if opcion == "1":

        print("\n--- REVISIÓN DEL PICO ---")
        estado = input("¿El pico está nuevo, usado o roto?: ")

        if estado == "nuevo":
            print("✅ El pico está listo para trabajar.")
        elif estado == "usado":
            print("⚠️ El pico necesita mantenimiento.")
        elif estado == "roto":
            print("❌ El pico no puede usarse.")
        else:
            print("⚠️ Estado no válido.")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n--- BÚSQUEDA DE MINERALES ---")
        mineral = input("¿Qué mineral buscas? (oro, plata o carbón): ")

        if mineral == "oro":
            print("💰 Encontraste oro.")
        elif mineral == "plata":
            print("🥈 Encontraste plata.")
        elif mineral == "carbón":
            print("⚫ Encontraste carbón.")
        else:
            print("❌ Mineral desconocido.")

    # OPCIÓN 3
    elif opcion == "3":

        print("\n--- ENERGÍA DEL MINERO ---")
        energia = int(input("Ingresa el nivel de energía (0-100): "))

        if energia >= 80:
            print("💪 El minero tiene mucha energía.")
        elif energia >= 50:
            print("🙂 El minero puede seguir trabajando.")
        elif energia >= 20:
            print("😓 El minero está cansado.")
        else:
            print("🛑 El minero debe descansar.")

    # OPCIÓN 4
    elif opcion == "4":

        print("\n👋 Gracias por usar la IA para Minero y Pico.")
        print("Programa finalizado.")
        print("Programa realizado por Samuel Nicola NL = 44 ")
        break

    # OPCIÓN INVÁLIDA
    else:
        print("❌ Opción no válida.")
```
